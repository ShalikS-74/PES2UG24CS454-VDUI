```python
import numpy as np,pandas as pd,matplotlib.pyplot as plt
sales=pd.read_csv('superstore.csv',encoding='latin-1'); sales['Order Date']=pd.to_datetime(sales['Order Date']); sales['Year']=sales['Order Date'].dt.year
```

```python
region_totals=sales.groupby('Region').Profit.sum().sort_values(); best_region=region_totals.idxmax(); fig,ax=plt.subplots(figsize=(7,5)); ax.axis('off'); ax.text(.5,.75,f'{best_region} is our top-profit region',ha='center',fontsize=22,fontweight='bold',color='#c0392b'); ax.text(.5,.55,f'Total profit: ${region_totals.max():,.0f}',ha='center',fontsize=13,color='#555'); ax.text(.5,.15,'Source: Northwind Superstore order data',ha='center',fontsize=8,color='lightgray'); plt.show()
cat_sales=sales.groupby('Product Category').Sales.sum(); colors={'Furniture':'#D9534F','Office Supplies':'#5cb85c','Technology':'#4C72B0'}; hatches={'Furniture':'//','Office Supplies':'xx','Technology':'..'}; fig,ax=plt.subplots();
for cat in cat_sales.index: ax.bar(cat,cat_sales[cat],color=colors[cat],hatch=hatches[cat],edgecolor='black',label=cat)
ax.legend(); plt.show()
```

```python
yearly_sales=sales.groupby('Year').Sales.sum(); y1,y2=yearly_sales.index[-2:]; v1,v2=yearly_sales[y1],yearly_sales[y2]; pct_change=(v2-v1)/v1*100; fig,axes=plt.subplots(1,2,figsize=(10,4)); axes[0].bar([str(y1),str(y2)],[v1,v2]); axes[0].set_ylim(0,max(v1,v2)*1.15); axes[0].set_title(f'Zero baseline: {pct_change:+.1f}%'); axes[1].bar([str(y1),str(y2)],[v1,v2],color='#c0392b'); axes[1].set_ylim(min(v1,v2)*.97,max(v1,v2)*1.03); axes[1].set_title(f'Truncated: {pct_change:+.1f}%'); plt.show()
```

```python
descriptive_version=f'Sales rose {pct_change:+.1f}% from {y1} to {y2}.'; interpretive_version=f'Sales rose {pct_change:+.1f}% from {y1} to {y2}, consistent with the strategy having a positive effect, although seasonality and market conditions have not been ruled out.'; print(descriptive_version); print(interpretive_version); plt.plot(yearly_sales.index,yearly_sales.values,'o-'); plt.annotate(interpretive_version,xy=(yearly_sales.index[-1],yearly_sales.values[-1]),xytext=(.02,.85),textcoords='axes fraction',arrowprops={'arrowstyle':'->'}); plt.show()
```
